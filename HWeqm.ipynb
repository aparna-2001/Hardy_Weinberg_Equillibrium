# Hardy-Weinberg Equilibrium Analysis

## Introduction

The Hardy-Weinberg principle states that allele and genotype frequencies in a population will remain constant from generation to generation in the absence of other evolutionary influences. These influences include:

- Mutations
- Gene flow (migration)
- Genetic drift
- Non-random mating
- Natural selection

This analysis calculates allele frequencies and tests whether a population is in Hardy-Weinberg Equilibrium (HWE).

## The Hardy-Weinberg Equation

**p² + 2pq + q² = 1**

Where:
- **p** = frequency of the dominant allele (A)
- **q** = frequency of the recessive allele (a)
- **p² (AA)** = frequency of homozygous dominant genotype
- **2pq (Aa)** = frequency of heterozygous genotype
- **q² (aa)** = frequency of homozygous recessive genotype

Also: **p + q = 1** (all allele frequencies must sum to 1)

---

## Step 1: Calculate Allele Frequencies

From the observed genotype counts, we calculate:
- Allele p frequency = (2 × number of AA + number of Aa) / (2 × total population)
- Allele q frequency = (2 × number of aa + number of Aa) / (2 × total population)

This is because each homozygous individual contributes 2 copies of that allele, while heterozygotes contribute 1 copy of each allele.

---

## Step 2: Hardy-Weinberg Equilibrium Test

### Interpretation:
- If p² + 2pq + q² = 1, the population **may be** in equilibrium
- However, this is just a mathematical check. Statistical tests (like Chi-square) are needed for rigorous testing

### What does HWE mean?
If a population is in Hardy-Weinberg Equilibrium:
- Allele frequencies are not changing over time
- The population is not evolving with respect to this gene
- All five HWE conditions are met (no mutation, random mating, no gene flow, infinite population size, no selection)

### What if it's NOT in equilibrium?
Deviation from HWE suggests one or more evolutionary forces are acting on the population.

---

## Step 3: Observed vs. Expected Genotype Frequencies

### Expected Frequencies (if in HWE):
- Expected AA = p² × total population
- Expected Aa = 2pq × total population  
- Expected aa = q² × total population

### Observed Frequencies:
These are the actual counts you provided from the population.

### Analysis:
Large differences between observed and expected values suggest the population is **not in Hardy-Weinberg Equilibrium**. This could indicate:
- Selection favoring or against certain genotypes
- Non-random mating patterns
- Population structure or genetic drift
- Migration or gene flow

---

## Step 4: Visualization of Hardy-Weinberg Curves

### Graph 1: Theoretical HWE Curves with User Data

This graph shows:
- **Blue curve (AA)**: Expected frequency of homozygous dominant (p²) across all possible p values
- **Green curve (Aa)**: Expected frequency of heterozygous (2pq) across all possible p values
- **Yellow curve (aa)**: Expected frequency of homozygous recessive (q²) across all possible p values
- **Red dots**: Your observed data points plotted on these theoretical curves

### Key Observations:
- The heterozygote frequency (Aa) peaks at p = 0.5 (when allele frequencies are equal)
- Homozygote frequencies (AA and aa) increase as their respective allele becomes more common
- If your red dots fall ON the curves, your population fits HWE expectations
- If they fall OFF the curves, there's deviation from HWE

### Graph 2: User Data Scatter Plot (p-axis)

This plot shows your three observed genotype frequencies plotted against the **dominant allele frequency (p)**.

It helps visualize:
- Where your population's genotype frequencies fall in relation to p
- Whether the pattern matches theoretical expectations

### Graph 3: User Data Scatter Plot (q-axis)

This plot shows your three observed genotype frequencies plotted against the **recessive allele frequency (q)**.

It provides the complementary view to the p-axis plot (since q = 1 - p).

---

## Step 5: Heterozygote Advantage/Disadvantage Analysis

### Comparing Heterozygote to Homozygote Frequencies

This bar chart shows the absolute differences:
- **|Aa - aa|**: Difference between heterozygote and homozygous recessive
- **|Aa - AA|**: Difference between heterozygote and homozygous dominant

### Interpretation:
- **Large differences** may indicate selection pressures
- If Aa frequency is much **higher** than both homozygotes: possible **heterozygote advantage** (e.g., sickle cell trait)
- If Aa frequency is much **lower** than expected: possible **heterozygote disadvantage** or inbreeding
- **Similar values** suggest balanced frequencies consistent with random mating

### For a Complete Analysis, Consider:
- **Chi-square test**: Statistical test for HWE (compare observed vs expected counts)
- **Sample size**: Larger populations give more reliable results
- **Biological context**: What evolutionary forces might be acting on this population?

### Potential Sources of Deviation:
- **Natural selection**: One genotype has higher fitness
- **Genetic drift**: Random changes in small populations
- **Migration**: Introduction of alleles from other populations
- **Non-random mating**: Inbreeding or assortative mating
- **Mutation**: Though usually negligible in one generation

### Real-World Applications:
- **Conservation biology**: Assessing genetic diversity in endangered species
- **Medical genetics**: Understanding disease allele frequencies in populations
- **Evolutionary biology**: Detecting selection and evolutionary change
- **Forensics**: Calculating genotype probabilities for DNA evidence

```python
from matplotlib import pyplot as plt
import numpy as np



Total_population = int( input('Enter the number of total population: '))
num_AA = int(input('Enter the number of homozygous dominant: '))
num_aa = int(input('Enter the number of homozygous recessive: '))
num_Aa = int(input('Enter the number of heterozygous: '))

freq_AA = num_AA/Total_population
freq_aa = num_aa/Total_population
freq_Aa = num_Aa/Total_population

p = freq_AA + 0.5*freq_Aa
q = freq_aa + 0.5*freq_Aa


homo_dom = p**2
homo_rec = q**2
hetero = 2 * p * q
ans = homo_dom + hetero + homo_rec
print(ans)

if ans == 1:
   print('The population is in Hardy Weinberg Equilibrium')
else:
   print('The population is not in Hardy Weinberg Equilibrium')


#observed and expected frequencies

observed_AA = num_AA
expected_AA = homo_dom * Total_population

observed_aa = num_aa
expected_aa = homo_rec * Total_population

observed_Aa = num_Aa
expected_Aa = hetero * Total_population

print(f"observed genotype frequency of homozygous dominant is {observed_AA} and expected genotype frequency is {expected_AA}" )

print(f"observed genotype frequency of homozygous recessive is {observed_aa} and expected genotype frequency is {expected_aa}" )

print(f"observed genotype frequency of heterozygous is {observed_Aa} and expected genotype frequency is {expected_Aa}" )



#visualisation

def graph(p, homo_dom, hetero, homo_rec):
    # theoretical curve
    p_range = np.linspace(0,1,100)
    q_range = 1 - p_range
    plt.plot(p_range, p_range**2, label='AA Expected Curve', linewidth=2, color='blue')
    plt.plot(p_range, 2*p_range*q_range, label='Aa Expected Curve', linewidth=2, color='green')
    plt.plot(p_range, q_range**2, label='aa Expected Curve', linewidth=2, color='yellow')

    # user data
    plt.scatter(p, homo_dom, color='red', s=70, label='User AA Value')
    plt.scatter(p, hetero, color='red', s=70, label='User Aa Value')
    plt.scatter(p, homo_rec, color='red', s=70, label='User aa Value')

    plt.xlabel('Allele frequency (p)')
    plt.ylabel('Genotype frequency')
    plt.legend()
    plt.show()

graph(p, homo_dom, hetero, homo_rec)


#scatterplot separate p

  # user data
def scatter_plot_p(p,homo_dom,hetero,homo_rec):
    plt.scatter(p, homo_dom, color='red', s=70, label='User AA Value')
    plt.scatter(p, hetero, color='green', s=70, label='User Aa Value')
    plt.scatter(p, homo_rec, color='yellow', s=70, label='User aa Value')

    plt.xlabel('Allele frequency (p)')
    plt.ylabel('Genotype frequency')
    plt.legend()
    plt.show()

scatter_plot_p(p,homo_dom,hetero,homo_rec)



#scatterplot separate q

  # user data
def scatter_plot_q(p,homo_dom,hetero,homo_rec):
    plt.scatter(q, homo_dom, color='red', s=70, label='User AA Value')
    plt.scatter(q, hetero, color='green', s=70, label='User Aa Value')
    plt.scatter(q, homo_rec, color='yellow', s=70, label='User aa Value')

    plt.xlabel('Allele frequency (q)')
    plt.ylabel('Genotype frequency')
    plt.legend()
    plt.show()

scatter_plot_q(q,homo_dom,hetero,homo_rec)

#ratio
# difference plot with labels

def diff_plot(diff_Aa_aa, diff_Aa_AA):
    labels = ['|Aa - aa|', '|Aa - AA|']
    values = [diff_Aa_aa, diff_Aa_AA]

    bars = plt.bar(labels, values)

    plt.ylabel('Absolute difference in genotype frequency')
    plt.title('Difference between heterozygote and homozygote frequencies')

    # add value labels on top of bars
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{height:.4f}',
            ha='center',
            va='bottom'
        )

    plt.show()

diff_Aa_aa = abs(hetero-homo_rec)
diff_Aa_AA = abs(hetero-homo_dom)

diff_plot(diff_Aa_aa, diff_Aa_AA)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Hardy-Weinberg Equilibrium equations:
# p + q = 1 (allele frequencies)
# p² + 2pq + q² = 1 (genotype frequencies)
# where p = freq of dominant allele, q = freq of recessive allele

# Create range of allele frequencies for p (from 0 to 1)
p = np.linspace(0, 1, 100)
q = 1 - p

# Calculate genotype frequencies
freq_AA = p**2      # Homozygous dominant
freq_Aa = 2*p*q     # Heterozygous
freq_aa = q**2      # Homozygous recessive

# Create the plot
plt.figure(figsize=(10, 6))

plt.plot(p, freq_AA, label='AA (p²)', linewidth=2, color='blue')
plt.plot(p, freq_Aa, label='Aa (2pq)', linewidth=2, color='green')
plt.plot(p, freq_aa, label='aa (q²)', linewidth=2, color='red')

plt.xlabel('Frequency of A allele (p)', fontsize=12)
plt.ylabel('Genotype Frequency', fontsize=12)
plt.title('Hardy-Weinberg Equilibrium', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.xlim(0, 1)
plt.ylim(0, 1)

# Add annotation for a specific example
example_p = 0.7
example_q = 0.3
plt.axvline(x=example_p, color='gray', linestyle='--', alpha=0.5)
plt.text(example_p + 0.02, 0.9, f'Example:\np = {example_p}\nAA = {example_p**2:.3f}\nAa = {2*example_p*example_q:.3f}\naa = {example_q**2:.3f}',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()

# Optional: Create a stacked area chart
plt.figure(figsize=(10, 6))

plt.fill_between(p, 0, freq_aa, label='aa (q²)', alpha=0.7, color='red')
plt.fill_between(p, freq_aa, freq_aa + freq_Aa, label='Aa (2pq)', alpha=0.7, color='green')
plt.fill_between(p, freq_aa + freq_Aa, 1, label='AA (p²)', alpha=0.7, color='blue')

plt.xlabel('Frequency of A allele (p)', fontsize=12)
plt.ylabel('Cumulative Genotype Frequency', fontsize=12)
plt.title('Hardy-Weinberg Equilibrium - Stacked View', fontsize=14, fontweight='bold')
plt.legend(fontsize=11, loc='center left')
plt.grid(True, alpha=0.3, axis='x')
plt.xlim(0, 1)
plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```
